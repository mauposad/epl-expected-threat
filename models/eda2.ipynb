# Unorganized Sandbox for Exploratory Data Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
import re
from collections import defaultdict
```

# Shape of the data

```python
df = pd.read_parquet("../data_pipeline/data/processed/season_12325_clean.parquet")
print(f"Number of columns: {len(df.columns)}")
print(df.columns[:30])
```

## Visualizing and grouping by theme

```python
# Define core themes manually (you can adjust this as needed)
themes = {
    "goals": ["goal"],
    "shots": ["shot"],
    "xg": ["xg"],
    "possession": ["possession"],
    "cards": ["card"],
    "fouls": ["foul"],
    "ppda": ["ppda"],
    "deep_passes": ["deep_pass"],
    "attack": ["attack", "attacks"],
    "defense": ["defense", "defence"],
    "passes": ["pass"],
    "crosses": ["cross"],
    "penalties": ["penalty"],
    "duels": ["duel"],
    "aerials": ["aerial"],
    "corners": ["corner"],
    "offside": ["offside"]
}

# Initialize
grouped_cols = defaultdict(list)

# Assign columns to themes based on keyword match
for col in df.columns:
    matched = False
    for theme, keywords in themes.items():
        if any(re.search(keyword, col, re.IGNORECASE) for keyword in keywords):
            grouped_cols[theme].append(col)
            matched = True
            break
    if not matched:
        grouped_cols["other"].append(col)  # fallback group

print(f"Number of themes: {len(grouped_cols)}")
print(grouped_cols.keys())
```

```python
for theme, cols in grouped_cols.items():
    print(f"{theme}: {len(cols)} columns")
```
